```python
sales = pd.DataFrame({
    "OrderID": [101,102,103,104,105,106,107,108],
    "Customer": ["Sankara","Priya","Rahul","Sneha","Priya","Rahul","Sankara","Neha"],
    "City": ["Delhi","Mumbai","Delhi","Pune","Mumbai","Delhi","Delhi","Pune"],
    "Category": ["Laptop","Phone","Phone","Laptop","Tablet","Laptop","Tablet","Phone"],
    "Amount": [80000,50000,35000,90000,25000,85000,30000,45000]
})

sales
```

```python
# Question 1
# Find the total sales (Amount) for each city.
sales.groupby('City')['Amount'].sum()
```

```python
# Question 3 (Multiple Aggregations)
# For each city, find:
# Total Sales
# Average Sales
# Maximum Sales
# Number of Orders
sales.groupby('City')['Amount'].agg(['sum', 'mean', 'max', 'count'])
```
